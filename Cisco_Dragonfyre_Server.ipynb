```python
# @title 🚀 BẤM VÀO ĐÂY ĐỂ BẬT SERVER CISCO DRAGONFYRE TỰ ĐỘNG 100%
# @markdown Bấm nút Play (▶️) bên dưới. Server sẽ tự động tải Server Pack 1.18GB và khởi chạy với 16GB RAM Colab Pro!

from google.colab import drive
import os, sys, glob, json, subprocess, time

print("=" * 65)
print("  🐉 KHỞI ĐỘNG MINECRAFT CISCO DRAGONFYRE [SERVER PACK CHÍNH THỨC]")
print("=" * 65)

# 1. Mount Google Drive để lưu trữ Map vĩnh viễn
print("\n[1/5] 📁 Kết nối Google Drive để lưu trữ Map vĩnh viễn...")
drive.mount('/content/drive')
DRIVE_SAVE_DIR = "/content/drive/MyDrive/Minecraft_Cisco_Server"
os.makedirs(DRIVE_SAVE_DIR, exist_ok=True)
os.makedirs(f"{DRIVE_SAVE_DIR}/world", exist_ok=True)
print(f"✅ Thư mục lưu Map vĩnh viễn: {DRIVE_SAVE_DIR}/world")

# 2. Cài Java 17 & công cụ mạng trên Colab
print("\n[2/5] ☕ Cài đặt Java 17 OpenJDK & công cụ...")
subprocess.run("apt-get update -qq && apt-get install -y -qq openjdk-17-jdk-headless wget curl jq unzip tmux", shell=True, check=True)
print("✅ Java 17 đã sẵn sàng!")

# 3. Tạo thư mục chạy server trên SSD siêu tốc Colab (/content/server)
SERVER_DIR = "/content/server"
os.makedirs(SERVER_DIR, exist_ok=True)
os.chdir(SERVER_DIR)

# Tải trọn bộ Server Pack chính thức (CARPG Dragonfyre V3C Beta Serv - 1.18GB)
ZIP_NAME = "CARPG_Dragonfyre_Server.zip"
if not os.path.exists("start.sh"):
    print("\n[3/5] ⚡ Đang tải trọn bộ Server Pack Cisco Dragonfyre (1.18GB) bằng mạng cáp quang Google...")
    print("⏳ Tốc độ mạng Colab cực nhanh, chỉ mất khoảng 10-20 giây...")
    DOWNLOAD_URL = "https://edge.forgecdn.net/files/8678/476/CARPG%20Dragonfyre%20V3C%20Beta%20Serv.zip"
    subprocess.run(f'wget -q --show-progress -c -O "{ZIP_NAME}" "{DOWNLOAD_URL}"', shell=True)
    print("📦 Đang giải nén hơn 250+ mods và thư viện trên SSD siêu tốc Colab...")
    subprocess.run(f'unzip -o -q "{ZIP_NAME}"', shell=True)
    subprocess.run(f'rm -f "{ZIP_NAME}"', shell=True)
    print("✅ Đã giải nén trọn bộ 250+ mods thành công!")
else:
    print("\n[3/5] ✅ Bộ Server Pack đã có sẵn trên máy chủ!")

# Cấp quyền thực thi cho start.sh
subprocess.run("chmod +x start.sh install_java.sh 2>/dev/null || true", shell=True)

# Liên kết (Symlink) thư mục world vào Google Drive để tự động lưu vĩnh viễn
if not os.path.islink(f"{SERVER_DIR}/world"):
    if os.path.exists(f"{SERVER_DIR}/world"):
        subprocess.run(f'cp -rn "{SERVER_DIR}/world/." "{DRIVE_SAVE_DIR}/world/" 2>/dev/null || true', shell=True)
        subprocess.run(f'rm -rf "{SERVER_DIR}/world"', shell=True)
    os.symlink(f"{DRIVE_SAVE_DIR}/world", f"{SERVER_DIR}/world")
    print("🔗 Đã liên kết thư mục World vào Google Drive -> Chơi đến đâu lưu thẳng vào Drive đến đó!")

# 4. Nạp cấu hình Acc Crack, Whitelist, Sethome, Login, 16GB RAM vào variables.txt & server.properties
print("\n[4/5] ⚙️ Nạp cấu hình: Crack, Whitelist 5 người, Sethome 5 home, Login, 16GB RAM...")
os.makedirs("mods", exist_ok=True)
os.makedirs("config", exist_ok=True)
os.makedirs("defaultconfigs", exist_ok=True)

with open("eula.txt", "w") as f:
    f.write("eula=true\n")

# Cấu hình variables.txt cấp 16GB RAM cho start.sh
if os.path.exists("variables.txt"):
    with open("variables.txt", "r") as f:
        v_lines = f.readlines()
    with open("variables.txt", "w") as f:
        for line in v_lines:
            if line.startswith("JAVA_ARGS="):
                f.write('JAVA_ARGS="-Xms12G -Xmx16G -XX:+UseG1GC -XX:+ParallelRefProcEnabled -XX:MaxGCPauseMillis=200 -XX:+UnlockExperimentalVMOptions -XX:+DisableExplicitGC -XX:+AlwaysPreTouch -XX:G1NewSizePercent=30 -XX:G1MaxNewSizePercent=40 -XX:G1ReservePercent=20 -XX:G1HeapWastePercent=5 -XX:G1MixedGCCountTarget=4 -XX:InitiatingHeapOccupancyPercent=15 -XX:G1MixedGCLiveThresholdPercent=90 -XX:G1RSetUpdatingPauseTimePercent=5 -XX:SurvivorRatio=32 -XX:+PerfDisableSharedMem -XX:MaxTenuringThreshold=1"\n')
            elif line.startswith("WAIT_FOR_USER_INPUT="):
                f.write('WAIT_FOR_USER_INPUT=false\n')
            else:
                f.write(line)

server_props = """# Minecraft server properties
enable-jmx-monitoring=false
rcon.port=25575
level-seed=
gamemode=survival
enable-command-block=true
enable-query=false
generator-settings={}
enforce-secure-profile=false
level-name=world
motd=\\u00a76\\u00a7lCisco Dragonfyre RPG \\u00a7a[Colab Pro 16GB RAM]
query.port=25565
pvp=true
generate-structures=true
max-chained-neighbor-updates=1000000
difficulty=normal
network-compression-threshold=256
max-tick-time=120000
require-resource-pack=false
use-native-transport=true
max-players=20
online-mode=false
enable-status=true
allow-flight=true
initial-disabled-packs=
broadcast-rcon-to-ops=true
view-distance=8
server-ip=
resource-pack-prompt=
allow-nether=true
server-port=25565
enable-rcon=false
sync-chunk-writes=true
op-permission-level=4
prevent-proxy-connections=false
hide-online-players=false
resource-pack=
entity-broadcast-range-percentage=100
simulation-distance=6
rcon.password=
player-idle-timeout=0
force-gamemode=false
rate-limit=0
hardcore=false
white-list=true
broadcast-console-to-ops=true
spawn-npcs=true
spawn-animals=true
log-ips=true
function-permission-level=2
initial-enabled-packs=vanilla
level-type=minecraft\\:normal
text-filtering-config=
spawn-monsters=true
enforce-whitelist=true
spawn-protection=0
resource-pack-sha1=
max-world-size=29999984
"""
with open("server.properties", "w") as f:
    f.write(server_props)

whitelist_data = [
  {"name": "wele_chanel", "uuid": "c30d9703-e18e-324b-a28a-6b801a1ef726"},
  {"name": "ZayvorMC", "uuid": "7d54e082-f79a-328d-afcb-1d5e3bbd8c8e"},
  {"name": "nomeme", "uuid": "f1be0c92-75d1-3ff8-9ca4-92795c64c782"},
  {"name": "blvibg", "uuid": "93e1b7f0-a681-37d4-8314-25e1df1dc982"},
  {"name": "NTT", "uuid": "92fe0a74-0f2c-3a83-a979-d58622839958"}
]
with open("whitelist.json", "w") as f:
    json.dump(whitelist_data, f, indent=2)

ftb_config = """{
  admin: {
  }
  misc: {
  }
  teleportation: {
    home: {
      cooldown: 10
      cooldown_bake: 0
      enabled: true
      max: 5
      warmup: 0
    }
    spawn: {
      cooldown: 10
      cooldown_bake: 0
      enabled: true
      warmup: 0
    }
    tpa: {
      cooldown: 10
      cooldown_bake: 0
      enabled: true
      warmup: 0
    }
    warp: {
      cooldown: 10
      cooldown_bake: 0
      enabled: true
      warmup: 0
    }
  }
}
"""
with open("defaultconfigs/ftbessentials-server.snbt", "w") as f:
    f.write(ftb_config)
with open("config/ftbessentials.snbt", "w") as f:
    f.write(ftb_config)

# Tải mod Login
if not os.path.exists("mods/loginsystem-1.7.jar"):
    subprocess.run('wget -q -O "mods/loginsystem-1.7.jar" "https://cdn.modrinth.com/data/u55HkO7b/versions/31K0L4hE/loginsystem-1.7.jar"', shell=True)

# 5. Cài đặt Playit tunnel và bật Server
print("\n[5/5] 🌐 Khởi tạo đường truyền mạng Playit.gg...")
subprocess.run("curl -sLo /usr/local/bin/playit https://github.com/playit-cloud/playit-agent/releases/download/v0.15.26/playit-linux-amd64 && chmod +x /usr/local/bin/playit", shell=True)
subprocess.run("tmux kill-session -t playit_tunnel 2>/dev/null || true", shell=True)
subprocess.run('tmux new-session -d -s playit_tunnel "/usr/local/bin/playit"', shell=True)
time.sleep(3)

print("\n" + "=" * 65)
print("🔗 LIÊN KẾT ĐƯỜNG TRUYỀN PLAYIT.GG:")
res = subprocess.run("tmux capture-pane -pt playit_tunnel | grep -E 'https://playit.gg/claim/' || true", shell=True, capture_output=True, text=True)
claim_url = res.stdout.strip()
if claim_url:
    print(f"👉 BẤM VÀO ĐÂY ĐỂ LẤY IP CỐ ĐỊNH CHO SERVER: {claim_url}")
else:
    print("✅ Playit.gg đã kết nối thành công với tài khoản của bạn!")
print("=" * 65)

print("\n🚀 ĐANG KHỞI CHẠY MINECRAFT FORGE 1.20.1 (16GB RAM COLAB PRO)...")
print("👉 Đang nạp 250+ mods RPG... Bạn sẽ thấy toàn bộ nhật ký Server hiện trực tiếp bên dưới!\n")

# Chạy start.sh trực tiếp với hiển thị thời gian thực (unbuffered)
proc = subprocess.Popen(["bash", "start.sh"], stdout=sys.stdout, stderr=sys.stderr)
proc.communicate()
```
